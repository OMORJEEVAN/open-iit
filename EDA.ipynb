first-pass outcome group

```python
def classify_outcome(disposition):

    if disposition in rpc_dispositions:
        return "RPC"

    elif disposition in [
        "third_party_contact",
        "third_party_ptp"
    ]:
        return "THIRD_PARTY"

    elif disposition in [
        "wrong_number",
        "invalid_number"
    ]:
        return "INVALID"

    elif disposition in [
        "no_answer",
        "call_rejected",
        "not_reachable",
        "switched_off"
    ]:
        return "UNRESOLVED"

    elif disposition == "language_barrier":
        return "OTHER"

    else:
        return "OTHER"


df_dial_attempts["outcome_group"] = (
    df_dial_attempts["disposition"]
    .apply(classify_outcome)
)

print(
    df_dial_attempts["outcome_group"]
    .value_counts()
)

print(
    df_dial_attempts["outcome_group"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

repeated behaviour of individual phones

```python
phone_history = (
    df_dial_attempts
    .groupby("phone_id")
    .agg(
        attempts=("attempt_id", "count"),
        rpc_count=("is_rpc", "sum"),
        third_party_count=(
            "outcome_group",
            lambda x: (x == "THIRD_PARTY").sum()
        ),
        invalid_count=(
            "outcome_group",
            lambda x: (x == "INVALID").sum()
        ),
        unresolved_count=(
            "outcome_group",
            lambda x: (x == "UNRESOLVED").sum()
        )
    )
)

phone_history["rpc_rate"] = (
    phone_history["rpc_count"]
    / phone_history["attempts"]
)

phone_history["third_party_rate"] = (
    phone_history["third_party_count"]
    / phone_history["attempts"]
)

phone_history["invalid_rate"] = (
    phone_history["invalid_count"]
    / phone_history["attempts"]
)

phone_history["unresolved_rate"] = (
    phone_history["unresolved_count"]
    / phone_history["attempts"]
)

print(phone_history.describe().round(3))
```

adding new feature to the dataset

```python
attempts = df_dial_attempts.copy()

attempts["attempt_ts"] = pd.to_datetime(
    attempts["attempt_ts"],
    errors="coerce"
)

attempts = attempts.sort_values(
    ["account_id", "phone_id", "attempt_ts"]
).reset_index(drop=True)

# This is the actual contact point being tested
attempts["contact_point_key"] = (
    attempts["account_id"].astype(str)
    + "_"
    + attempts["phone_id"].astype(str)
)

# Primary outcome
attempts["is_rpc"] = (
    attempts["disposition"]
    .astype(str)
    .str.startswith("rpc_")
)

# Safety / invalid-contact outcome
attempts["is_unsafe"] = attempts["disposition"].isin([
    "third_party_contact",
    "third_party_ptp",
    "wrong_number",
    "invalid_number"
])

print("Rows:", len(attempts))
print("RPC:", attempts["is_rpc"].sum())
print("Unsafe:", attempts["is_unsafe"].sum())

print("\nRPC rate:")
print(attempts["is_rpc"].mean())

print("\nUnsafe rate:")
print(attempts["is_unsafe"].mean())
```

```python
attempts.head(5)
```

```python
g = attempts.groupby(
    "contact_point_key",
    sort=False
)

# Number of attempts BEFORE the current attempt
attempts["prior_attempts"] = (
    g.cumcount()
)

# Number of previous RPCs
attempts["prior_rpc"] = (
    g["is_rpc"].cumsum()
    - attempts["is_rpc"].astype(int)
)

# Number of previous unsafe outcomes
attempts["prior_unsafe"] = (
    g["is_unsafe"].cumsum()
    - attempts["is_unsafe"].astype(int)
)

# Historical rates
attempts["prior_rpc_rate"] = np.where(
    attempts["prior_attempts"] > 0,
    attempts["prior_rpc"] / attempts["prior_attempts"],
    0.0
)

attempts["prior_unsafe_rate"] = np.where(
    attempts["prior_attempts"] > 0,
    attempts["prior_unsafe"] / attempts["prior_attempts"],
    0.0
)

print(
    attempts[
        [
            "contact_point_key",
            "attempt_ts",
            "prior_attempts",
            "prior_rpc",
            "prior_unsafe",
            "prior_rpc_rate",
            "prior_unsafe_rate"
        ]
    ].head(20)
)
```

```python
attempts["previous_disposition"] = (
    g["disposition"].shift(1)
)

attempts["previous_network_response"] = (
    g["network_response"].shift(1)
)

attempts["previous_hangup_by"] = (
    g["hangup_by"].shift(1)
)

attempts["previous_attempt_ts"] = (
    g["attempt_ts"].shift(1)
)

attempts["days_since_previous_attempt"] = (
    attempts["attempt_ts"]
    - attempts["previous_attempt_ts"]
).dt.total_seconds() / (24 * 3600)

attempts["days_since_previous_attempt"] = (
    attempts["days_since_previous_attempt"]
    .fillna(-1)
)

print(
    attempts[
        [
            "contact_point_key",
            "attempt_ts",
            "previous_disposition",
            "previous_network_response",
            "days_since_previous_attempt"
        ]
    ].head(20)
)
```
